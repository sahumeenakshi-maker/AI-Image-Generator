```python
all_results = data['result']['data']['results']
image_links = []

for item in all_results:
    original = item.get("origin")
    thumbnail = item.get("thumb")
    
    image_links.append(original)
    image_links.append(thumbnail)

print(f"Found {len(image_links)} links in total.")
print("^" * 40)
print(f"Found {(image_links)} links in total.")
```
